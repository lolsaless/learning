## 1. 필요한 도구 불러오기
표를 다루는 도구(`pandas`)와 그래프 도구(`matplotlib`)를 불러옵니다.

```python
import pandas as pd
import matplotlib.pyplot as plt
import random

print("준비 완료!")
```

## 2. 예시 측정 데이터 만들기
실제 측정기처럼 **5분 간격으로 2시간** 동안 측정한 CO2 값을 만듭니다. (가짜 데이터입니다)

```python
random.seed(1)  # 매번 같은 예시 데이터가 나오도록 고정

시각 = pd.date_range("2025-04-08 10:00", periods=25, freq="5min")

측정 = pd.DataFrame({
    "측정시각": list(시각) * 2,
    "측정지점": ["1층 교실"] * 25 + ["2층 교실"] * 25,
    "CO2_ppm": [random.randint(550, 800) for _ in range(25)]
             + [random.randint(850, 1250) for _ in range(25)],
})

측정.head(10)   # 앞의 10줄만 보기
```

## 3. 측정지점별 평균 구하고 판정하기
기준(예시): **CO2 1,000ppm 이상이면 부적합**

```python
기준 = 1000

요약 = 측정.groupby("측정지점")["CO2_ppm"].agg(["mean", "max", "min"]).round(1)
요약.columns = ["평균", "최대", "최소"]
요약["판정"] = ["부적합" if 평균 >= 기준 else "적합" for 평균 in 요약["평균"]]

요약
```

## 4. 그래프 그리기
시간에 따라 CO2가 어떻게 변했는지 봅니다. 빨간 점선이 기준선입니다.

> 그래프 글자는 한글 글꼴 설정 없이도 보이도록 영어로 적었습니다. (한글 글꼴 설정은 31회에서 배웁니다)

```python
plt.figure(figsize=(9, 4))

for 지점, 영문 in [("1층 교실", "1F classroom"), ("2층 교실", "2F classroom")]:
    자료 = 측정[측정["측정지점"] == 지점]
    plt.plot(자료["측정시각"], 자료["CO2_ppm"], marker="o", label=영문)

plt.axhline(기준, color="red", linestyle="--", label="Standard 1,000 ppm")
plt.title("Indoor CO2 (5-minute interval)")
plt.ylabel("CO2 (ppm)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
