### 문제 1
변수 `place`, `item`, `value`를 만들어 `3층 회의실 HCHO 72.5 ㎍/㎥` 를 f-string으로 출력하세요.

### 문제 3

### 문제 6
리스트의 **처음 3개 평균**과 **마지막 3개 평균**을 각각 소수 1자리로 출력하세요. (기대: 1010.0 / 1084.3)

### 문제 7
딕셔너리 `mw`에 `"SO2": 64.07`을 추가하고, **분자량이 40보다 큰 항목 이름**만 출력하세요. (기대: CO2, NO2, SO2)

```python
mw = {"CO2": 44.01, "HCHO": 30.03, "CO": 28.01, "NO2": 46.01}
mw["SO2"] = 64.07
for name, m in mw.items():
    if m > 40:
        print(name)
```

### 문제 8
`samples`에서 `place`가 `"2층"`인 시료의 `value`를 모두 리스트로 모으세요. (기대: [1052.8, 41])

```python
samples = [
    {"place": "1층", "item": "CO2", "value": 667.4},
    {"place": "2층", "item": "CO2", "value": 1052.8},
    {"place": "2층", "item": "PM2.5", "value": 41},
    {"place": "3층", "item": "CO2", "value": 554.9},
]
print([s["value"] for s in samples if s["place"] == "2층"])
```

### 문제 9

### 문제 10
폼알데하이드 값이 80 이상 → `"부적합"`, 72 이상(기준의 90 %) → `"주의"`, 그 외 `"적합"`. `hcho` = 85, 75, 60으로 확인하세요.

```python
hcho = 75
if hcho >= 80:
    print("부적합")
elif hcho >= 80 * 0.9:
    print("주의")
else:
    print("적합")
```

### 문제 12
측정값 중 **기준(1000) 이상인 값의 위치**(1부터 셈)를 모두 출력하세요. (기대: 2, 3, 4, 5)

### 문제 13
두 리스트를 짝지어 `1층교실: 적합` 처럼 출력하세요. (기준 1000)

```python
places = ["1층교실", "2층교실", "3층사무실"]
averages = [650.2, 1047.2, 558.8]
for p, a in zip(places, averages):
    print(f"{p}: {'부적합' if a >= 1000 else '적합'}")
```

### 문제 14
`while`로: 압력 600 psi에서 **매주 10 %씩 줄어들 때**, 100 psi 아래가 되는 것은 몇 주 차인가요? (기대: 18주)

```python
pressure = 600
week = 0
while pressure >= 100:
    pressure *= 0.9
    week += 1
print(week, "주", round(pressure, 1))
```

### 문제 15
리스트 컴프리헨션 한 줄로, ppm 값들을 mg/㎥(정수)로 바꾼 새 리스트를 만드세요. (분자량 44.01, 몰부피 24.45)

```python
co2 = [930, 1012, 1088]
print([round(v * 44.01 / 24.45) for v in co2])
```
