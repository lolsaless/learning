## 1. 결측값(NA) 다루기 — `20230202.R`
```r
z <- c(1,2,NA,4,5,NA,9)
is.na(z)
sum(is.na(z))
mean(z)                 # NA
mean(z, na.rm = TRUE)   # 4.2
```
| R | 파이썬 (기본 문법) |
|---|---|
| `c(1, 2, NA)` | `[1, 2, None]` — 값이 없음은 `None` |
| `is.na(z)` | `[v is None for v in z]` |
| `sum(is.na(z))` | `sum(v is None for v in z)` |
| `mean(z, na.rm=TRUE)` | 결측을 빼고 평균 |

> 💡 4단계 pandas에서는 결측을 `NaN`으로 표시하고 `.isna()`, `.mean()`(자동으로 결측 제외)을 씁니다. R과 아주 비슷합니다.

````python
z = [1, 2, None, 4, 5, None, 9]

print([v is None for v in z])
print("결측 개수:", sum(v is None for v in z))

valid = [v for v in z if v is not None]
print("결측 제외 평균:", round(sum(valid) / len(valid), 1))
````

### 🧪 연습 1
측정 기록에 결측(`None`)이 섞여 있습니다. **결측 개수**, **결측 비율(%)**, **결측을 뺀 평균**을 출력하세요.

````python
pm10 = [42, None, 55, 61, None, 39, 52, None]
missing = sum(v is None for v in pm10)
valid = [v for v in pm10 if v is not None]
print("결측:", missing, "개 /", round(missing / len(pm10) * 100, 1), "%")   # 3개 / 37.5 %
print("평균:", round(sum(valid) / len(valid), 1))                          # 49.8
````

## 2. 데이터 프레임 기초 — `20230202.R`
R 코드는 `mpg`(자동차 연비 예제 데이터)로 행·열 수, 열 이름, 앞부분 보기, 새 데이터 프레임 만들기를 연습합니다.
```r
nrow(mpg); ncol(mpg); names(mpg); head(mpg, 2)
thempg <- data.frame("제조사" = x, 모델 = y, 연비 = z)
```

파이썬 기본 문법으로는 **딕셔너리의 리스트**(14회)로 같은 일을 할 수 있습니다. (4단계에서는 pandas `DataFrame`)

## 3. 리스트와 벡터 연산 — `20230203.R`
```r
list4 <- list(thempg, 1:10, list3)
names(list4) <- c("data.frame", "vector", "list")
a <- c(1:10); b <- c(11:20)
a + b        # 같은 위치끼리 더하기 → 12 14 16 ...
```
| R | 파이썬 |
|---|---|
| 이름 붙은 `list(...)` | 딕셔너리 `{"vector": ..., "list": ...}` |
| `list4[[2]]` | `list4["vector"]` 또는 (리스트라면) `x[1]` |
| `1:10` | `list(range(1, 11))` |
| `a + b` (벡터 덧셈) | `[x + y for x, y in zip(a, b)]` |

````python
a = list(range(1, 11))
b = list(range(11, 21))
print(a + b)                               # ⚠️ 파이썬 리스트의 + 는 '이어 붙이기'!
print([x + y for x, y in zip(a, b)])       # R의 a + b 와 같은 결과
````

> ⚠️ **R과 파이썬이 다르게 동작하는 대표적인 곳**입니다. R의 `a + b`는 위치끼리 더하지만, 파이썬 리스트의 `a + b`는 **이어 붙입니다.** R 코드를 옮길 때 AI도 가끔 이 실수를 합니다. (4단계 pandas의 열끼리 `+`는 R처럼 위치끼리 더합니다)

### 🧪 연습 2
측정값(ppm) 리스트와 보정계수 리스트가 있습니다. **위치끼리 곱한** 보정값 리스트를 만드세요. (R이라면 `values * factors`)

````python
values = [612, 845, 1020]
factors = [1.02, 0.98, 1.01]
print([round(v * f, 1) for v, f in zip(values, factors)])    # [624.2, 828.1, 1030.2]
````

## 4. 행렬 — `20230203.R`
```r
A <- matrix(1:10, nrow = 5)   # 5행 2열, 열 방향으로 채움
t(A)                          # 전치 (행↔열)
```
파이썬 기본 문법에서는 **리스트의 리스트**로 표현합니다. 행렬 계산이 많아지면 `numpy` 라이브러리를 씁니다(5단계 이후).

````python
# R의 matrix(1:10, nrow=5) 는 '열 방향'으로 채움 → 1~5가 첫 열, 6~10이 둘째 열
A = [[1, 6],
     [2, 7],
     [3, 8],
     [4, 9],
     [5, 10]]
print("2행 1열:", A[1][0])               # R의 A[2, 1]  (R은 1부터!)
At = [list(row) for row in zip(*A)]      # 전치 t(A)
print("전치:", At)
````

받은 코드를 아래에 붙여 넣고 실행한 뒤, **R과 다르게 동작할 수 있는 부분**이 있는지 AI에게 다시 물어보세요.

````python
# (예시) AI 답을 붙여 넣고 실행한 결과를 확인하는 칸입니다.
# 20230215.R 은 nycflights13 데이터로 그룹별 요약을 연습하는 코드입니다.
# 아래는 '그룹별 평균'을 기본 문법으로 하는 예입니다.
flights = [
    {"carrier": "AA", "delay": 12}, {"carrier": "AA", "delay": -3},
    {"carrier": "UA", "delay": 25}, {"carrier": "UA", "delay": 5}, {"carrier": "UA", "delay": None},
]
groups = {}
for f in flights:
    if f["delay"] is not None:                           # R의 na.rm = TRUE
        groups.setdefault(f["carrier"], []).append(f["delay"])
for carrier, delays in groups.items():
    print(carrier, round(sum(delays) / len(delays), 1))
````
