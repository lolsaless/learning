### 문제 1
두 수의 **상대 차이(%)** = |a − b| ÷ 평균 × 100 을 소수 1자리로 돌려주는 함수 `rpd(a, b)`를 만드세요. 중복 시료 분석의 정밀도(RPD)에 씁니다. `rpd(10.2, 9.8)` → 4.0

```python
def rpd(a, b):
    return round(abs(a - b) / ((a + b) / 2) * 100, 1)

print(rpd(10.2, 9.8))
```

### 문제 2
측정값 리스트와 기준을 받아 **(초과 개수, 초과율 %)** 를 돌려주는 함수 `over_rate(values, limit=1000)`를 만드세요. 초과율은 소수 1자리. `[930, 1012, 1088, 998]` → (2, 50.0)

```python
def over_rate(values, limit=1000):
    n = sum(1 for v in values if v >= limit)
    return n, round(n / len(values) * 100, 1)

print(over_rate([930, 1012, 1088, 998]))
```

### 문제 3
함수 `grade(value, bounds, names)`: 구간 경계 리스트와 등급 이름 리스트로 등급을 돌려주세요. `grade(36, [15, 35, 75], ["좋음","보통","나쁨","매우나쁨"])` → `"나쁨"` (힌트: 경계를 차례로 보며 `value <= 경계`인 첫 위치)

```python
def grade(value, bounds, names):
    for i, b in enumerate(bounds):
        if value <= b:
            return names[i]
    return names[-1]

names = ["좋음", "보통", "나쁨", "매우나쁨"]
print(grade(36, [15, 35, 75], names), grade(15, [15, 35, 75], names), grade(80, [15, 35, 75], names))
```

### 문제 8
`to_number(text)`를 만드세요: 공백·쉼표 제거 후 float, 실패하면 None. `[" 1,020 ", "N/A", "45.2"]` → [1020.0, None, 45.2]

```python
def to_number(text):
    try:
        return float(text.strip().replace(",", ""))
    except ValueError:
        return None

print([to_number(t) for t in [" 1,020 ", "N/A", "45.2"]])
```

### 문제 11
`practice_files` 폴더를 만들고, `day1.csv`~`day3.csv` 세 파일에 `CO2_ppm` 열과 값 3개씩을 쓰세요. 값은 `[600, 650, 700]`, `[900, 1000, 1100]`, `[500, 520, 540]`.

```python
from pathlib import Path
import csv
folder = Path("practice_files")
folder.mkdir(exist_ok=True)
data = {"day1": [600, 650, 700], "day2": [900, 1000, 1100], "day3": [500, 520, 540]}
for name, values in data.items():
    with open(folder / f"{name}.csv", "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f)
        w.writerow(["CO2_ppm"])
        for v in values:
            w.writerow([v])
print(sorted(p.name for p in folder.glob("*.csv")))
```

### 문제 12
`practice_files`의 CSV마다 평균을 구해 `파일이름: 평균` 으로 출력하세요. (기대: day1 650.0, day2 1000.0, day3 520.0)

```python
for path in sorted(folder.glob("*.csv")):
    with open(path, encoding="utf-8") as f:
        vals = [int(r["CO2_ppm"]) for r in csv.DictReader(f)]
    print(f"{path.stem}: {sum(vals) / len(vals)}")
```

### 문제 13
문제 12의 결과를 `practice_files/summary.csv`에 **엑셀에서 한글이 깨지지 않게** 저장하세요. 열: 파일, 평균, 판정(기준 1000 이상 부적합)

```python
rows = []
for path in sorted(folder.glob("day*.csv")):
    with open(path, encoding="utf-8") as f:
        vals = [int(r["CO2_ppm"]) for r in csv.DictReader(f)]
    avg = sum(vals) / len(vals)
    rows.append({"파일": path.name, "평균": avg, "판정": "부적합" if avg >= 1000 else "적합"})
with open(folder / "summary.csv", "w", newline="", encoding="utf-8-sig") as f:
    w = csv.DictWriter(f, fieldnames=["파일", "평균", "판정"])
    w.writeheader()
    w.writerows(rows)
print((folder / "summary.csv").read_text(encoding="utf-8-sig"))
```

### 문제 15
AI와 함께: 측정값 리스트를 받아 **이동평균**(연속 3개 평균) 리스트를 돌려주는 함수를 AI에게 만들게 하고, `[930, 1012, 1088, 1045, 1120]` → `[1010.0, 1048.3, 1084.3]` 가 나오는지 확인하세요.

```python
def moving_average(values, n=3):
    return [round(sum(values[i:i + n]) / n, 1) for i in range(len(values) - n + 1)]

print(moving_average([930, 1012, 1088, 1045, 1120]))
```
