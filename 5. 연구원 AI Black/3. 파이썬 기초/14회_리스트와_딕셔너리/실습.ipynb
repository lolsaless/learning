## 1. 리스트 — 순서가 있는 값의 줄
대괄호 `[ ]` 안에 쉼표로 나열합니다. 엑셀의 **한 열**과 비슷합니다.

```python
co2 = [930, 1012, 1088, 1045, 1120, 998, 1061]
print(co2)
print("측정 횟수:", len(co2))
```

### 꺼내기 — 번호(인덱스)는 0부터!
| 표현 | 뜻 | 결과 |
|---|---|---|
| `co2[0]` | 첫 번째 | 930 |
| `co2[2]` | 세 번째 | 1088 |
| `co2[-1]` | 마지막 | 1061 |
| `co2[1:4]` | 두 번째부터 네 번째까지 (4번 **앞**까지) | [1012, 1088, 1045] |
| `co2[:3]` | 처음 3개 | [930, 1012, 1088] |

```python
print("첫 측정:", co2[0])
print("마지막 측정:", co2[-1])
print("처음 3개:", co2[:3])
print("2~4번째:", co2[1:4])
```

### 요약 함수

```python
print("합계:", sum(co2))
print("최대:", max(co2))
print("최소:", min(co2))
print("평균:", round(sum(co2) / len(co2), 1))
print("작은 순:", sorted(co2))
print("큰 순:", sorted(co2, reverse=True))
```

## 2. 딕셔너리 — 이름표가 붙은 값
중괄호 `{ }` 안에 `키: 값` 을 나열합니다. **찾아보기 표**(VLOOKUP)와 비슷합니다.

> ⚠️ 아래 기준값은 **수업용 예시**입니다. 실제 기준은 법령 원문으로 확인하세요. (06회)

```python
limits = {
    "PM10": 75,
    "PM2.5": 35,
    "CO2": 1000,
    "HCHO": 80,
}
print(limits)
print("CO2 기준:", limits["CO2"])
```

| 하는 일 | 코드 |
|---|---|
| 값 꺼내기 | `limits["CO2"]` |
| 없을 수도 있을 때 | `limits.get("라돈", "없음")` → 키가 없으면 "없음" |
| 추가·바꾸기 | `limits["CO"] = 10` |
| 키 목록 / 값 목록 | `limits.keys()`, `limits.values()` |
| 키와 값 함께 | `limits.items()` (16회 반복문에서 사용) |
| 들어 있나? | `"CO2" in limits` |

```python
limits["CO"] = 10                     # 새 항목 추가
print(limits.get("라돈", "기준 없음"))  # 없는 키 → 오류 대신 기본값
print("항목들:", list(limits.keys()))
print("CO2 들어 있나?", "CO2" in limits)
```

## 3. 딕셔너리의 리스트 — 표 만들기
시료 하나 = 딕셔너리 하나, 시료 여러 개 = **딕셔너리의 리스트**. 엑셀의 **행 여러 줄**과 같습니다.
4단계의 pandas 표(DataFrame)가 바로 이 모양에서 출발합니다.

```python
samples = [
    {"id": "S-01", "place": "1층 교실", "item": "CO2", "value": 667.4},
    {"id": "S-02", "place": "2층 교실", "item": "CO2", "value": 1052.8},
    {"id": "S-03", "place": "3층 사무실", "item": "CO2", "value": 554.9},
]
print("시료 수:", len(samples))
print("두 번째 시료:", samples[1])
print("두 번째 시료의 측정값:", samples[1]["value"])
```
