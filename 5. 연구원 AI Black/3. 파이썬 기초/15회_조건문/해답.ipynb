# 15회. 조건문 — 기준 초과 판정기

**오늘의 목표**
1. `if` · `elif` · `else`로 **조건에 따라 다른 일**을 하게 한다.
2. `and` · `or` · `not`으로 조건을 조합한다.
3. **들여쓰기**가 파이썬 문법의 일부라는 것을 안다.
4. 측정값 **판정기**와 **등급 분류기**를 만든다.

> ⚠️ 이 노트북의 기준값·등급 구간은 **수업용 예시**입니다. 실제 업무에는 법령·고시 원문 값을 쓰세요.

> ✅ **해답 노트북** — 연습 문제의 예시 답안이 채워져 있습니다.

## 1. if — 만약 ~라면
```python
if 조건:
    조건이 참일 때 할 일      ← 공백 4칸 들여쓰기 (Tab 키)
```
- 조건 뒤에 **콜론(`:`)**, 다음 줄은 **들여쓰기**. 들여쓴 줄들이 "if에 속한 덩어리"입니다.

````python
co2 = 1052.8
limit = 1000

if co2 >= limit:
    print("⚠️ 기준 초과")
    print("시설 담당자에게 통보 필요")
print("판정 끝")      # 들여쓰지 않은 줄은 조건과 관계없이 항상 실행
````

## 2. else — 아니면

````python
co2 = 667.4
if co2 >= limit:
    result = "부적합"
else:
    result = "적합"
print(f"CO2 {co2} ppm → {result}")
````

### 🧪 연습 1
`hcho = 72.5`, 기준 `hcho_limit = 80` 일 때, 기준 **이상**이면 `"부적합"`, 아니면 `"적합"`을
변수 `result`에 담아 `폼알데하이드 72.5 ㎍/㎥ → 적합` 처럼 출력하세요. 값을 `85`로 바꿔서도 실행해 보세요.

````python
hcho = 72.5
hcho_limit = 80
if hcho >= hcho_limit:
    result = "부적합"
else:
    result = "적합"
print(f"폼알데하이드 {hcho} ㎍/㎥ → {result}")
````

## 3. elif — 여러 구간으로 나누기 (등급)
위에서부터 차례로 검사해서 **처음 참인 곳 하나만** 실행합니다.

| PM10 (㎍/㎥, 예시) | 등급 |
|---|---|
| 0~30 | 좋음 |
| 31~80 | 보통 |
| 81~150 | 나쁨 |
| 151 이상 | 매우나쁨 |

````python
pm10 = 95

if pm10 <= 30:
    grade = "좋음"
elif pm10 <= 80:
    grade = "보통"
elif pm10 <= 150:
    grade = "나쁨"
else:
    grade = "매우나쁨"

print(f"PM10 {pm10} → {grade}")
````

> 💡 두 번째 줄을 `elif 31 <= pm10 <= 80:` 처럼 쓰지 않아도 되는 이유: 첫 번째 조건(`<= 30`)이 거짓이어야 여기까지 오므로, 이미 31 이상이라는 뜻입니다.

### 🧪 연습 2
PM2.5 등급(예시)을 판정하세요. **0~15 좋음, 16~35 보통, 36~75 나쁨, 76 이상 매우나쁨**
`pm25` 값을 12, 35, 36, 80으로 바꿔 가며 확인하세요.

````python
pm25 = 36
if pm25 <= 15:
    grade = "좋음"
elif pm25 <= 35:
    grade = "보통"
elif pm25 <= 75:
    grade = "나쁨"
else:
    grade = "매우나쁨"
print(f"PM2.5 {pm25} → {grade}")    # 나쁨
````

## 4. and · or · not — 조건 조합
| 연산 | 뜻 | 예 |
|---|---|---|
| `A and B` | 둘 다 참 | 온도 18~28 ℃ 사이: `temp >= 18 and temp <= 28` (또는 `18 <= temp <= 28`) |
| `A or B` | 하나라도 참 | CO2 **또는** 폼알데하이드 초과 → 부적합 |
| `not A` | 반대 | `not over` |

````python
co2, hcho = 845, 92
co2_limit, hcho_limit = 1000, 80

if co2 >= co2_limit or hcho >= hcho_limit:
    print("시설 판정: 부적합 (하나 이상 기준 초과)")
else:
    print("시설 판정: 적합")

temp = 31.5
if 18 <= temp <= 28:
    print("측정 환경 온도 적정")
else:
    print("측정 환경 온도 범위 밖 → 기록 필요")
````

### 🧪 연습 3
측정값이 **유효한지** 먼저 검사하는 코드를 만드세요.
- 값이 0보다 작으면 `"측정 오류(음수)"`
- 값이 5000보다 크면 `"측정 오류(범위 초과) - 장비 확인"`
- 그 외에는 기준 1000으로 `"적합"` / `"부적합"` 판정

`co2` 를 -5, 7200, 850, 1200 으로 바꿔 가며 확인하세요.

````python
co2 = 7200
if co2 < 0:
    result = "측정 오류(음수)"
elif co2 > 5000:
    result = "측정 오류(범위 초과) - 장비 확인"
elif co2 >= 1000:
    result = "부적합"
else:
    result = "적합"
print(co2, "→", result)
````

## 5. 딕셔너리와 함께 쓰기
14회의 기준값 딕셔너리를 쓰면, **항목이 바뀌어도 같은 코드**로 판정할 수 있습니다.

````python
limits = {"PM10": 75, "PM2.5": 35, "CO2": 1000, "HCHO": 80}   # 수업용 예시
sample = {"place": "2층 교실", "item": "PM2.5", "value": 41}

item = sample["item"]
if item in limits:
    if sample["value"] >= limits[item]:
        result = "부적합"
    else:
        result = "적합"
else:
    result = "기준 없음"

print(f'{sample["place"]} {item} {sample["value"]} → {result}')
````

### 🧪 연습 4 — AI와 함께
위 셀의 `if` 안에 `if`가 들어간 모양(중첩)을 AI에게 보여 주고
> "이 코드를 들여쓰기가 덜 깊게, 초보자가 읽기 쉽게 바꿔 줘. 결과는 같아야 해."

라고 요청하세요. 받은 코드를 아래에 붙여 넣고, `sample`의 항목을 `"라돈"`, `"CO2"`(값 900)로 바꿔 결과가 같은지 확인하세요.

````python
sample = {"place": "2층 교실", "item": "PM2.5", "value": 41}
item = sample["item"]
if item not in limits:
    result = "기준 없음"
elif sample["value"] >= limits[item]:
    result = "부적합"
else:
    result = "적합"
print(f'{sample["place"]} {item} {sample["value"]} → {result}')
````

## 정리
| 문법 | 모양 |
|---|---|
| 기본 | `if 조건:` → 들여쓴 줄 |
| 아니면 | `else:` |
| 여러 구간 | `elif 조건:` (위에서부터, 처음 참인 곳 하나만) |
| 조합 | `and`, `or`, `not`, `18 <= temp <= 28` |
| 주의 | 콜론 `:`, 들여쓰기 4칸, `=`(담기) vs `==`(같은가) |
