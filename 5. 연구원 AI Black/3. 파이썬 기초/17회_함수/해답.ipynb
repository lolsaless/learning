## 1. 왜 함수인가?
15·16회에서 판정 코드를 여러 번 썼습니다. 같은 코드를 **복사·붙여넣기**하면
- 기준이 바뀌면 모든 곳을 고쳐야 하고
- 한 곳만 고치고 다른 곳을 빠뜨리는 실수가 생깁니다.

→ 자주 쓰는 코드에 **이름을 붙여 두고** 필요할 때 부르는 것이 **함수**입니다. 엑셀의 `AVERAGE()`, `IF()`가 바로 함수입니다.

```python
def 함수이름(입력1, 입력2):
    할 일
    return 결과
```

````python
def judge(value, limit):
    """측정값이 기준 이상이면 '부적합', 아니면 '적합'을 돌려준다."""
    if value >= limit:
        return "부적합"
    return "적합"

print(judge(1052.8, 1000))
print(judge(667.4, 1000))
print(judge(72.5, 80))
````

- `def` 줄: 함수 **만들기** (이때는 아무것도 실행되지 않음)
- `judge(1052.8, 1000)`: 함수 **부르기(호출)** → `value=1052.8`, `limit=1000`이 들어가 실행
- `return`: 결과를 **돌려주고 함수를 끝냄**. 그래서 위 코드에서 `else`가 없어도 됩니다.

### return vs print
| | `print` | `return` |
|---|---|---|
| 하는 일 | 화면에 **보여 주기만** | 결과를 **돌려줘서 다른 곳에서 쓰게** |
| 결과를 변수에 담기 | 안 됨 (`None`) | 됨 |

````python
def show_only(value, limit):
    print("부적합" if value >= limit else "적합")

a = show_only(1052.8, 1000)     # 화면에는 보이지만
print("a에 담긴 값:", a)          # None — 돌려받은 게 없음

b = judge(1052.8, 1000)
print("b에 담긴 값:", b)
````

### 🧪 연습 1
첨가 회수율(%)을 계산하는 함수 `recovery(measured, spiked)`를 만드세요.
- 회수율 = 측정값 ÷ 첨가량 × 100, **소수 1자리**로 돌려주기
- `recovery(9.2, 10)` → `92.0`, `recovery(0.47, 0.5)` → `94.0`

````python
def recovery(measured, spiked):
    """첨가 회수율(%)을 소수 1자리로 돌려준다."""
    return round(measured / spiked * 100, 1)

print(recovery(9.2, 10))     # 92.0
print(recovery(0.47, 0.5))   # 94.0
````

## 2. 기본값이 있는 매개변수
자주 쓰는 값은 **기본값**으로 정해 두면, 부를 때 생략할 수 있습니다.

````python
def ppm_to_mgm3(ppm, mw, molar_volume=24.45):
    """기체 농도 ppm → mg/㎥ (기본: 25 ℃, 1기압의 몰부피 24.45 L/mol)."""
    return ppm * mw / molar_volume

print(round(ppm_to_mgm3(1000, 44.01)))                        # 기본 몰부피
print(round(ppm_to_mgm3(1000, 44.01, molar_volume=22.4)))     # 0 ℃ 기준으로 바꿔서
````

### 🧪 연습 2
`judge`를 고쳐 `limit`의 기본값을 `1000`으로 하는 `judge_co2(value, limit=1000)`를 만드세요.
`judge_co2(1052.8)`, `judge_co2(1052.8, limit=1200)` 의 결과를 확인하세요.

````python
def judge_co2(value, limit=1000):
    return "부적합" if value >= limit else "적합"

print(judge_co2(1052.8))               # 부적합
print(judge_co2(1052.8, limit=1200))   # 적합
````

## 3. 여러 값 돌려주기
`return a, b, c` 처럼 쓰면 여러 값을 **한 번에** 돌려줄 수 있습니다. (튜플)
값이 많으면 **딕셔너리**로 돌려주는 것이 읽기 좋습니다.

````python
def summarize(values):
    """측정값 리스트의 요약(개수, 평균, 최대, 최소)을 딕셔너리로 돌려준다."""
    return {
        "count": len(values),
        "mean": round(sum(values) / len(values), 1),
        "max": max(values),
        "min": min(values),
    }

co2_1f = [612, 645, 701, 688, 720]
co2_2f = [930, 1012, 1088, 1045, 1120]

s1 = summarize(co2_1f)
s2 = summarize(co2_2f)
print(s1)
print("2층 평균:", s2["mean"])
````

## 4. 함수 조립하기 — 작은 함수를 모아 큰 일

````python
def report_line(place, values, limit=1000):
    """측정 지점 하나의 결과를 한 줄 문장으로 만든다."""
    s = summarize(values)
    result = judge(s["mean"], limit)
    return f"{place}: {s['count']}회 측정, 평균 {s['mean']} ppm (최대 {s['max']}) → {result}"

print(report_line("1층 교실", co2_1f))
print(report_line("2층 교실", co2_2f))
````

### 🧪 연습 3
폼알데하이드 **㎍/㎥ → ppm** 환산 함수 `ugm3_to_ppm(ug, mw=30.03, molar_volume=24.45)`를 만들고,
`[45.2, 62.1, 85.3]` 각각을 ppm으로 바꿔 **소수 4자리**로 출력하세요. (13회 연습 4 참고)

````python
def ugm3_to_ppm(ug, mw=30.03, molar_volume=24.45):
    """㎍/㎥ → ppm"""
    return ug / 1000 * molar_volume / mw

for ug in [45.2, 62.1, 85.3]:
    print(f"{ug} ㎍/㎥ = {ugm3_to_ppm(ug):.4f} ppm")
````

### 🧪 연습 4
여러 지점의 측정값이 딕셔너리로 있습니다. 위에서 만든 `report_line`을 **반복문 안에서** 불러 모든 지점의 보고 문장을 출력하세요.

````python
data = {
    "1층 교실": [612, 645, 701, 688, 720],
    "2층 교실": [930, 1012, 1088, 1045, 1120],
    "3층 사무실": [512, 548, 601, 577, 630],
}
for place, values in data.items():
    print(report_line(place, values))
````

## 5. 설명문(docstring)과 AI
함수 첫 줄의 `"""설명"""`은 **docstring**입니다. VS Code에서 함수 이름에 마우스를 올리면 이 설명이 보입니다.

### 🧪 연습 5 — AI와 함께
아래 함수를 AI에게 보여 주고 **"이 함수에 초보자도 이해할 수 있는 한글 docstring과 주석을 달아 줘. 매개변수와 반환값 설명도 넣어 줘."** 라고 요청하세요.
받은 코드를 붙여 넣고 `grade_pm10(95)`로 실행해 보세요.
```python
def grade_pm10(v):
    if v <= 30: return "좋음"
    if v <= 80: return "보통"
    if v <= 150: return "나쁨"
    return "매우나쁨"
```

````python
def grade_pm10(v):
    """
    PM10 농도(㎍/㎥)를 받아 예보 등급(수업용 예시 구간)을 돌려준다.

    매개변수
        v : PM10 농도 (㎍/㎥)
    반환값
        "좋음" / "보통" / "나쁨" / "매우나쁨"
    """
    if v <= 30:            # 0~30
        return "좋음"
    if v <= 80:            # 31~80
        return "보통"
    if v <= 150:           # 81~150
        return "나쁨"
    return "매우나쁨"      # 151 이상

print(grade_pm10(95))
````

## 정리
| 문법 | 예 |
|---|---|
| 만들기 | `def judge(value, limit):` |
| 돌려주기 | `return "부적합"` (`print`와 다름!) |
| 기본값 | `def f(ppm, mw, molar_volume=24.45):` |
| 여러 값 | `return {"mean": …, "max": …}` |
| 설명 | `"""docstring"""` |

> 💡 좋은 함수는 **한 가지 일**만 합니다. `summarize`(요약), `judge`(판정), `report_line`(문장 만들기)처럼 나눠 두면 고치기 쉽고 재사용하기 좋습니다.
