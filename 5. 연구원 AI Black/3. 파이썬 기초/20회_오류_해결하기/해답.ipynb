## 1. 오류 메시지 읽는 법

````python
def mean(values):
    return sum(values) / len(values)

def report(place, values):
    return f"{place} 평균 {mean(values):.1f}"

print(report("2층 교실", []))     # 측정값이 하나도 없는 경우
````

### 읽는 순서: **맨 아래 → 위로**
1. **맨 아래 줄** = 오류 종류 + 설명 → `ZeroDivisionError: division by zero` (0으로 나눔)
2. **바로 위** = 오류가 **실제로 난 줄** → `return sum(values) / len(values)`
3. **더 위** = 그 줄까지 **어떻게 왔는지** (report → mean)

→ 해석: "측정값이 빈 리스트라 개수가 0이고, 0으로 나눠서 오류". 코드가 틀린 게 아니라 **데이터에 문제**가 있는 경우입니다.

### 🧪 연습 2-5 — 논리 오류 (오류 메시지 없음!)
아래 코드는 오류 없이 실행되지만 **평균이 틀립니다.** 기대 결과는 `평균 1039.0`, `초과 4회` 입니다. 무엇이 문제일까요?

````python
values = [930, 1012, 1088, 1045, 1120]
total = 0          # 누적 변수는 반복 '전에' 0으로 (16회)
over = 0
for v in values:
    total = total + v
    if v > 1000:
        over += 1
print("평균", total / len(values), "/ 초과", over, "회")
````

> 💡 **논리 오류가 가장 위험합니다.** 오류 메시지가 없어서 틀린 줄 모르고 보고서에 들어갈 수 있습니다.
> 그래서 **손으로 계산할 수 있는 작은 데이터**로 먼저 결과를 확인하는 습관이 중요합니다. 이 예에서는 930 + 1012 + 1088 + 1045 + 1120 = 5195, ÷ 5 = **1039.0** 을 계산기로 확인할 수 있습니다.

## 4. try / except — 예상되는 문제를 미리 처리
측정 기록에는 `"N/A"`, 빈칸, 오타(`"9OO"` — 숫자 0 대신 영문 O) 같은 값이 섞여 있습니다. 하나 때문에 전체가 멈추면 안 됩니다.

```python
try:
    문제가 생길 수 있는 코드
except 오류종류:
    문제가 생겼을 때 할 일
```

````python
raw = ["612", "845", "N/A", "1020", "", "9OO", "1088"]

values = []
skipped = []
for text in raw:
    try:
        values.append(int(text))
    except ValueError:
        skipped.append(text)

print("숫자로 바꾼 값:", values)
print("건너뛴 값:", skipped)
print("평균:", round(sum(values) / len(values), 1))
````

> ⚠️ `except:` (오류 종류 없이) 로 **모든 오류를 삼키지 마세요.** 진짜 문제까지 숨겨집니다. **예상한 오류(`ValueError`)만** 잡고, 건너뛴 값은 **기록**해서 사람이 확인하게 합니다.

### 🧪 연습 3
측정값 문자열 리스트에서 **소수도 처리**하도록 `float()`으로 바꾸고, 바꿀 수 없는 값은 건너뛰되 **몇 번째 값인지**(1부터) 함께 출력하세요.
`평균` 계산은 `values`가 비어 있으면 `"유효한 측정값 없음"`을 출력하도록 하세요.

````python
raw = ["45.2", "62.1", "-", "85.3", "불검출", "78.9"]
values = []
for i, text in enumerate(raw, 1):
    try:
        values.append(float(text))
    except ValueError:
        print(f"{i}번째 값 '{text}' 건너뜀")

if values:
    print("평균:", round(sum(values) / len(values), 1))   # 67.9
else:
    print("유효한 측정값 없음")
````

## 5. 디버깅 4단계
| 단계 | 방법 |
|---|---|
| ① 읽기 | 오류 메시지 **맨 아래 줄**과 **오류 난 줄** 확인 |
| ② 확인 | 의심되는 변수를 `print(변수, type(변수))`로 찍어 보기 |
| ③ 줄이기 | 데이터를 **2~3개로 줄여** 손으로 계산 가능한 상태에서 실행 |
| ④ 묻기 | AI에게 **코드 + 오류 메시지 전체 + 하려던 일 + 환경**을 함께 |

### AI에게 오류 물어보는 템플릿 (`공통자료/프롬프트_템플릿.md` 3-2)
```
파이썬 코드를 실행했더니 오류가 났어. 원인과 해결 방법을 알려 줘.
[하려던 일] ...
[실행한 코드] (붙여넣기)
[오류 메시지 전체] (붙여넣기 — 일부만 붙이지 말 것)
[환경] VS Code 주피터 노트북, Python 3.13, Mac/Windows
초보자도 이해하게 원인을 먼저 설명하고, 고친 코드 전체를 보여 줘.
```

### 🧪 연습 4 — AI와 함께
아래 코드는 오류가 납니다. 위 템플릿으로 AI에게 물어보고, **AI의 설명을 한 줄로 요약**한 뒤 고친 코드를 실행하세요.

````python
# AI 설명 요약: 값이 문자열('667.4')이라 숫자 1000과 크기 비교를 할 수 없음 (TypeError) → float()으로 변환
def judge(value, limit=1000):
    return "부적합" if value >= limit else "적합"

readings = {"1층": "667.4", "2층": "1052.8"}
for place, value in readings.items():
    print(place, judge(float(value)))
````

---
## 정리
- 오류 메시지는 **맨 아래 → 위로** 읽는다. 맨 아래 = 무슨 오류, 그 위 = 어느 줄.
- **TypeError·ValueError**는 대부분 "숫자처럼 보이는 문자열" 문제 (13회).
- **논리 오류**는 메시지가 없다 → 작은 데이터로 **검산**. 기대 결과도 검산!
- `try/except`는 **예상한 오류만** 잡고, 건너뛴 것은 기록한다.
- AI에게는 **코드 + 오류 전체 + 하려던 일 + 환경**을 함께.
