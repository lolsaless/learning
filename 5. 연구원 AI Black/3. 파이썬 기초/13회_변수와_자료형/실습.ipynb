## 1. print와 주석

## 2. 변수 — 값에 이름표 붙이기
`=` 는 "같다"가 아니라 **"오른쪽 값을 왼쪽 이름에 담는다"** 입니다.

```
co2 = 1036      →   [co2] 라는 이름표가 붙은 상자에 1036을 넣음
```

````python
place = "2층 교실"
co2 = 1036
limit = 1000

print(place, "CO2:", co2, "ppm / 기준:", limit, "ppm")
````

````python
# 변수에 새 값을 넣으면 이전 값은 사라진다
co2 = 1036
co2 = 980
print(co2)
````

### 변수 이름 짓기 규칙
| 규칙 | 좋은 예 | 안 되는 예 |
|---|---|---|
| 영어 소문자 + 숫자 + 밑줄(`_`) | `co2_avg`, `pm10_1f` | `co2 avg` (공백), `1f_pm10` (숫자로 시작) |
| 뜻이 보이게 | `limit`, `place_name` | `a`, `x1` |
| 파이썬 예약어 피하기 | `max_value` | `if`, `for`, `print`(덮어쓰면 고장) |

> 한글 변수 이름(`측정값 = 10`)도 동작은 하지만, 이 과정에서는 **영어 이름**을 씁니다. AI가 만든 코드와 모양이 같아 읽기 쉽고, 다른 도구와도 잘 맞습니다.

## 3. 자료형 — 값의 종류
| 자료형 | 이름 | 예 | 쓰임 |
|---|---|---|---|
| `int` | 정수 | `1036`, `-5` | 개수, 측정 횟수 |
| `float` | 실수(소수) | `45.2`, `0.08` | 농도, 온도 |
| `str` | 문자열 | `"2층 교실"`, `"CO2"` | 이름, 메모, 접수번호 |
| `bool` | 참·거짓 | `True`, `False` | 판정 결과 |

`type()`으로 자료형을 확인할 수 있습니다.

````python
print(type(1036))
print(type(45.2))
print(type("2층 교실"))
print(type(1036 >= 1000))
````

## 4. 계산하기
| 연산 | 기호 | 예 | 결과 |
|---|---|---|---|
| 더하기 / 빼기 | `+` `-` | `1036 - 1000` | `36` |
| 곱하기 | `*` | `612 * 2` | `1224` |
| 나누기 | `/` | `7 / 2` | `3.5` (항상 실수) |
| 몫 / 나머지 | `//` `%` | `125 // 60`, `125 % 60` | `2`, `5` (125분 = 2시간 5분) |
| 거듭제곱 | `**` | `10 ** 3` | `1000` |

````python
total = 612 + 845 + 1020
count = 3
average = total / count

print("합계:", total)
print("평균:", average)
print("평균(소수 1자리):", round(average, 1))
````

````python
# 측정 시간 125분은 몇 시간 몇 분?
minutes = 125
print(minutes // 60, "시간", minutes % 60, "분")
````

## 5. 문자열 다루기 맛보기
- `+` 로 이어 붙이기, `*` 로 반복하기
- **f-string**: 따옴표 앞에 `f`를 붙이고 `{변수}`를 넣으면 값이 들어감 → 결과 문장을 만들 때 가장 많이 씀

````python
item = "CO2"
unit = "ppm"
print(item + " (" + unit + ")")
print("-" * 20)

place = "2층 교실"
average = 1036.27
print(f"{place}의 {item} 평균은 {average} {unit}입니다.")
print(f"{place}의 {item} 평균은 {average:.1f} {unit}입니다.")   # :.1f → 소수 1자리
````

## 6. 형 변환 — 숫자처럼 보이는 문자열 조심!
엑셀·CSV·입력창에서 읽은 값은 **문자열**인 경우가 많습니다. 문자열은 계산이 되지 않습니다.

````python
a = "612"
b = "845"
print(a + b)                 # 문자열끼리 + 는 이어 붙이기 → '612845'
print(int(a) + int(b))       # 정수로 바꾸면 계산 → 1457
print(float("45.2") * 2)     # 소수는 float
print(str(1036) + " ppm")    # 숫자를 글자로
````

## 8. 업무 예제 — 단위 환산 (ppm ↔ mg/㎥)
기체 농도의 부피 단위(ppm)와 질량 단위(mg/㎥)는 **분자량**과 **몰부피**로 환산합니다.
(25 ℃, 1기압에서 몰부피 약 24.45 L/mol 기준)

```
mg/㎥ = ppm × 분자량 ÷ 24.45
ppm   = mg/㎥ × 24.45 ÷ 분자량
```
| 물질 | 분자량 (g/mol) |
|---|---|
| 이산화탄소 CO2 | 44.01 |
| 폼알데하이드 HCHO | 30.03 |

````python
molar_volume = 24.45      # L/mol (25 ℃, 1 atm)
mw_co2 = 44.01            # 이산화탄소 분자량

co2_ppm = 1000
co2_mg = co2_ppm * mw_co2 / molar_volume
print(f"CO2 {co2_ppm} ppm = {co2_mg:.0f} mg/㎥")
````
