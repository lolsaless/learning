## 1. for — 하나씩 꺼내서 반복
```python
for 값 in 리스트:
    값으로 할 일        ← 들여쓰기
```

````python
co2 = [930, 1012, 1088, 1045, 1120, 998, 1061]

for value in co2:
    print(value, "ppm")
````

````python
# 반복 + 조건문 = 일괄 판정
limit = 1000
for value in co2:
    if value >= limit:
        print(value, "→ 초과")
    else:
        print(value, "→ 적합")
````

## 2. 누적 계산 — 세기와 더하기
반복 **전에** 0으로 시작하는 변수를 만들고, 반복 **안에서** 키워 갑니다.

````python
over_count = 0           # 초과 횟수 (반복 전에 0)
total = 0                # 합계

for value in co2:
    total = total + value          # total += value 와 같음
    if value >= limit:
        over_count += 1

print("합계:", total, "/ 평균:", round(total / len(co2), 1))
print("초과 횟수:", over_count, "/", len(co2))
````

## 3. range, enumerate, zip
| 도구 | 하는 일 | 예 |
|---|---|---|
| `range(5)` | 0,1,2,3,4 | 5번 반복 |
| `range(1, 6)` | 1~5 | 회차 번호 |
| `enumerate(리스트, 1)` | (번호, 값) 함께 | 몇 번째 측정인지 |
| `zip(A, B)` | 두 리스트를 나란히 | 지점 이름과 측정값 짝짓기 |

````python
for i in range(1, 4):
    print(f"{i}회차 측정")

for number, value in enumerate(co2, 1):
    print(f"{number}번째 측정: {value} ppm")
````

````python
places = ["1층 교실", "2층 교실", "3층 사무실"]
averages = [667.4, 1052.8, 554.9]

for place, avg in zip(places, averages):
    result = "부적합" if avg >= limit else "적합"     # 한 줄 if (조건 표현식)
    print(f"{place:<8} {avg:>7.1f} ppm  {result}")
````

> 💡 `result = "부적합" if avg >= limit else "적합"` 은 if/else 4줄을 한 줄로 쓴 것입니다. AI가 자주 쓰는 모양이라 읽을 수 있으면 좋습니다.
> `{place:<8}`은 왼쪽 정렬 8칸, `{avg:>7.1f}`는 오른쪽 정렬 7칸 소수 1자리 — 표처럼 줄을 맞출 때 씁니다.

## 4. 딕셔너리 반복 — 여러 항목 판정

````python
limits = {"PM10": 75, "PM2.5": 35, "CO2": 1000, "HCHO": 80}        # 수업용 예시
result_1f = {"PM10": 48, "PM2.5": 38, "CO2": 845, "HCHO": 62}

for item, value in result_1f.items():
    status = "부적합" if value >= limits[item] else "적합"
    print(f"{item:6} {value:>6}  (기준 {limits[item]})  {status}")
````

## 5. while — 조건이 될 때까지
**몇 번 반복할지 모를 때** 씁니다. 예: 헬륨 봄베 압력이 출구압까지 떨어지려면 **몇 주**?
(22회 미니 프로젝트의 예고편입니다)

> ⚠️ 조건이 영원히 참이면 끝나지 않습니다(무한 반복). 그럴 땐 노트북 위쪽 **[중단]** 버튼 또는 터미널에서 `Ctrl + C`.

````python
pressure = 600        # 현재 봄베 압력 (psi)
cutoff = 80           # 이 압력 아래로는 사용 불가
weekly_drop = 47.5    # 한 주에 떨어지는 압력 (가정)

week = 0
while pressure > cutoff:
    print(f"+{week}주: {pressure:.1f} psi")
    pressure = pressure - weekly_drop
    week += 1

print(f"→ 약 {week}주 차에 교체 필요")
````

## 정리
| 문법 | 쓰임 |
|---|---|
| `for v in 리스트:` | 값 하나씩 꺼내 반복 |
| `count = 0` → 반복 안에서 `count += 1` | 개수 세기, 합계 |
| `range`, `enumerate`, `zip`, `.items()` | 번호, 짝짓기, 딕셔너리 반복 |
| `while 조건:` / `break` | 조건이 될 때까지 / 찾으면 멈춤 |
| `[v for v in A if 조건]` | 한 줄로 걸러 모으기 |
