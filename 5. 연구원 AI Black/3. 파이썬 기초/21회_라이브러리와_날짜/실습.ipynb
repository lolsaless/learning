## 1. import — 도구 상자 꺼내기
| 모양 | 뜻 | 쓰는 법 |
|---|---|---|
| `import math` | math 상자 통째로 | `math.ceil(2.1)` |
| `import statistics as st` | 별명 붙이기 | `st.mean(...)` |
| `from datetime import datetime` | 상자 안의 도구 하나만 | `datetime(...)` |

> 표준 라이브러리는 **설치 없이** 씁니다. pandas처럼 따로 설치하는 것은 "외부 라이브러리"입니다 (11회 `pip install`).

```python
import math
import statistics as st

values = [930, 1012, 1088, 1045, 1120, 998, 1061]

print("평균:", round(st.mean(values), 1))
print("중앙값:", st.median(values))
print("표준편차:", round(st.stdev(values), 1))       # 표본 표준편차 (엑셀 STDEV.S)
print("올림:", math.ceil(1036.3), "/ 내림:", math.floor(1036.3))
print("제곱근:", math.sqrt(2))
```

## 2. 날짜와 시간 — datetime
### 문자열 → 날짜 (`strptime`)
`"2025-04-16 09:40"`은 그냥 **글자**라서 빼기를 할 수 없습니다. **형식(format)** 을 알려 주고 날짜로 바꿉니다.

| 기호 | 뜻 | 예 |
|---|---|---|
| `%Y` | 연도 4자리 | 2025 |
| `%m` | 월 2자리 | 04 |
| `%d` | 일 2자리 | 16 |
| `%H` / `%M` | 시 / 분 | 09 / 40 |

```python
from datetime import datetime, timedelta

start = datetime.strptime("2025-04-16 09:40", "%Y-%m-%d %H:%M")
end = datetime.strptime("2025-04-16 11:50", "%Y-%m-%d %H:%M")
print(start, type(start))

duration = end - start                 # 날짜끼리 빼면 '기간'(timedelta)
print("기간:", duration)
print("분:", duration.total_seconds() / 60)
print("시간:", round(duration.total_seconds() / 3600, 2))
```

### 날짜 → 문자열 (`strftime`), 날짜 더하기 (`timedelta`)

```python
today = datetime(2025, 4, 8, 9, 0)
print(today.strftime("%Y년 %m월 %d일 %H시 %M분"))

next_check = today + timedelta(days=90)          # 90일 뒤 재측정
print("재측정 예정:", next_check.strftime("%Y-%m-%d"))

weekdays = ["월", "화", "수", "목", "금", "토", "일"]
print("요일:", weekdays[next_check.weekday()])   # weekday(): 월=0 … 일=6
```

## 3. 출장기록 — 출장 시간 계산 (18회 이어서)
18회에서 출장기간을 4칸(출발일, 출발시간, 도착일, 도착시간)으로 나눴습니다.
이제 **출장 시간**을 계산합니다. 시간이 `--:--`(빈칸)이면 **날짜만으로 일수**를 계산합니다.

```python
trips = [
    {"name": "직원A", "start_date": "2025-04-08", "start_time": "09:30", "end_date": "2025-04-08", "end_time": "17:00"},
    {"name": "직원B", "start_date": "2025-04-09", "start_time": "",      "end_date": "2025-04-09", "end_time": ""},
    {"name": "직원C", "start_date": "2025-04-10", "start_time": "13:00", "end_date": "2025-04-11", "end_time": "12:00"},
]

def trip_length(t):
    """출장 시간을 '○시간 ○분' 또는 (시간이 없으면) '○일'로 돌려준다."""
    if t["start_time"] and t["end_time"]:              # 빈 문자열은 거짓(False)으로 취급
        s = datetime.strptime(t["start_date"] + " " + t["start_time"], "%Y-%m-%d %H:%M")
        e = datetime.strptime(t["end_date"] + " " + t["end_time"], "%Y-%m-%d %H:%M")
        minutes = int((e - s).total_seconds() // 60)
        return f"{minutes // 60}시간 {minutes % 60}분"
    s = datetime.strptime(t["start_date"], "%Y-%m-%d")
    e = datetime.strptime(t["end_date"], "%Y-%m-%d")
    return f"{(e - s).days + 1}일 (시간 기록 없음)"

for t in trips:
    print(t["name"], trip_length(t))
```

> 💡 `if t["start_time"]:` — 빈 문자열 `""`, `0`, 빈 리스트 `[]`, `None`은 조건에서 **거짓**으로 취급됩니다. "값이 있으면"을 짧게 쓰는 관례입니다.

## 4. 처리 일수 — 평일만 세기
접수일부터 결과 확인일까지 **며칠 걸렸는지**, 그리고 그중 **평일(월~금)은 며칠**인지 셉니다. (공휴일은 이번에는 고려하지 않음)

```python
received = datetime.strptime("2025-04-08", "%Y-%m-%d")
confirmed = datetime.strptime("2025-04-17", "%Y-%m-%d")

print("달력 일수:", (confirmed - received).days, "일")

workdays = 0
day = received
while day < confirmed:
    day = day + timedelta(days=1)
    if day.weekday() < 5:           # 월(0)~금(4)
        workdays += 1
print("평일 기준:", workdays, "일")
```
