# 나의 첫 주피터 노트북

> **10회 VS Code** 실습 파일입니다.

주피터 노트북(`.ipynb`)은 **설명 글**과 **코드**, **실행 결과**를 한 문서에 함께 담습니다.
분석 과정을 기록하고 다른 사람에게 보여 주기 좋아서, 4~5단계 데이터 분석은 대부분 노트북으로 합니다.

## 칸(셀)의 두 종류
| 셀 | 모양 | 하는 일 |
|---|---|---|
| **마크다운 셀** | 지금 이 글 | 설명을 적음 (제목, 목록, 표) |
| **코드 셀** | 회색 칸 | 파이썬 코드를 적고 실행 |

## 실행 방법
- 코드 셀을 클릭하고 **Shift + Enter** → 실행하고 다음 셀로 이동
- 셀 왼쪽의 **▶ 버튼**을 눌러도 됩니다.
- 처음 실행할 때 오른쪽 위에서 **커널(파이썬)** 을 고르라고 하면 강의노트의 안내를 따르세요.

## 1. 첫 코드 실행

````python
print("안녕하세요, 주피터 노트북!")
````

## 2. 값을 이름에 담기 (변수)
`=` 는 "같다"가 아니라 **"오른쪽 값을 왼쪽 이름에 담는다"** 는 뜻입니다.

````python
place = "2층 교실"
co2 = 1036

print(place, "의 CO2는", co2, "ppm 입니다.")
````

## 3. 셀의 마지막 줄은 자동으로 보여 준다
노트북에서는 `print` 없이 **마지막 줄의 값**이 바로 출력됩니다.

````python
values = [930, 1012, 1088, 1045, 1120, 998, 1061]
sum(values) / len(values)
````

## 4. 앞 셀의 값을 이어서 쓰기
노트북은 **위에서 실행한 셀의 값을 기억**합니다. 위 셀을 실행하지 않고 이 셀을 실행하면 오류가 납니다. (직접 확인해 보세요)

````python
limit = 1000
average = sum(values) / len(values)

if average >= limit:
    print("판정: 부적합")
else:
    print("판정: 적합")
````

## 5. 직접 해 보기

### 5-1. 아래 셀의 측정값을 바꿔 실행해 보세요
`values`의 숫자를 모두 900 이하로 바꾸면 판정이 어떻게 바뀌나요?

````python
values = [930, 1012, 1088, 1045, 1120, 998, 1061]   # ← 숫자를 바꿔 보세요
average = sum(values) / len(values)
print("평균:", round(average, 1))
print("판정:", "부적합" if average >= limit else "적합")
````
